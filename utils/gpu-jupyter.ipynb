# Get Jupyter Notebook to use GPU
- By default my Jupyter notebooks aren't able to use my GPU locally

## CUDA
- Need to make sure `NVIDIA Graphics Drivers`, `Nvidia CUDA Toolkit`, and `cuDNN` are installed/compatible
- I've seen conda/miniconda recommended for a one-step solution, but I prefer to avoid those
- if I want to use the GPU locally without that, looks like I'll need to enable on a library-by-library basis

## PyTorch

### PyTorch Start Locally Guide
- this [Start Locally guide from PyTorch](https://pytorch.org/get-started/locally/)
- you select stuff about your system and it gives you a command to run
- on Windows, run `nvidia-smi` in Command Prompt to find out CUDA version supported by driver (top right of table)
- it gave me the command:

```bash
pip3 install torch torchvision --index-url https://download.pytorch.org/whl/cu130
```

### Using UV with PyTorch guide
- oh neat, UV has it's own [Using UV with PyTorch Guide](https://docs.astral.sh/uv/guides/integration/pytorch/#installing-pytorch), should've looked for this in the first place
- the difficulty is that there are different versions of PyTorch for CPU/CUDA/different versions of CUDA, and PyPI only has the CPU-only wheels for Windows.
- followed along with the instructions to modify `pyproject.toml` for it to pull the correct version of PyTorch
- after that, notebooks were able to use the GPU, drastically speeding up training time.

### Check if PyTorch is using the GPU

````python
# Check if PyTorch can use GPU
import torch
print(f"PyTorch version: {torch.version.__version__}")
cuda_available = torch.cuda.is_available()
print(f"CUDA available: { torch.cuda.is_available()}")
if cuda_available:
    print(f"Using device: {torch.cuda.get_device_name(0)}")
````
