# Scikit-Learn

scikit-learn focuses on traditional ML methods like classification, clustering, and clustering, and is easier to gets tarted with than the Deep Learning-focused libraries like PyTorch, TensorFlow, and Keras.

## [Gentle Intro to Scikit-Learn](https://machinelearningmastery.com/a-gentle-introduction-to-scikit-learn-a-python-machine-learning-library/)
- a Machine Learning Mastery article with a great overview of [scikit-learn](https://scikit-learn.org/stable/)
- initially developed by David Cournapeau for a 2007 Google "summer of code" project
- at the time of the article, there were 30+ active contributors - compared to massive efforts like the Meta-produced pytorch, it suggests that it's smaller-scale but more open and user-focused.
- it mentions that scikit-learn's [Getting Started](https://scikit-learn.org/stable/getting_started.html) page is a decent place to ... start.
